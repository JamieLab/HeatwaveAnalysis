```python
# Section to run the Taylor Decomposition of air-sea CO2 fluxes
import matplotlib.pyplot as plt
import PyCO2SYS as pyco2
import numpy as np
import pandas as pd
import matplotlib.transforms
font = {'weight' : 'normal',
        'size'   : 14}
matplotlib.rc('font', **font)
def gas_transfer(wind2):
    kw = 0.271 * wind2
    return np.array(kw) #cm hr-1

def schmidt(temp):
    sch = (2116.8
        - 136.25 * temp
        + 4.7353 * temp ** 2
        - 0.092307 * temp ** 3
        + 0.0007555 * temp ** 4)
    return np.array(sch)

def solubility(temp,sal):
    temp = np.array(temp); sal = np.array(sal)
    temp = temp+273.15 # Convert temperature to Kelvin
    
    sol = (-58.0931
        + 90.5069 * (100.0 / temp)
        + 22.2940 * np.log(temp / 100.0)
        + sal * (
            0.027766
            - 0.025888 * (temp / 100.0)
            + 0.0050578 * (temp / 100.0) ** 2
            )
        )
    sol = np.exp(sol)
    return np.array(sol/1000) #mol L –1 atm–1 -> mol m-3 atm-1

def flux(fco2sw,fco2atm,temp,sal,wind):
    fco2sw = np.array(fco2sw)
    fco2atm = np.array(fco2atm)
    
    kw = gas_transfer(np.array(wind))
    sch = schmidt(temp)
    sol = solubility(temp,sal)
    fluxes = kw * sol * (fco2sw - fco2atm) / 3600 /100  # cm hr-1 * mol m-3 atm-1 * uatm * cm->m * hr to seconds
    return -fluxes

def get_cumulated_array(data,**kwargs):
    cum = data.clip(**kwargs)
    cum = np.cumsum(cum, axis=0)
    d = np.zeros(np.shape(data))
    d[1:] = cum[:-1]
    return d  
def orderOfMagnitude(number):
    return np.floor(np.math.log(number, 10))

cols = ['#332288','#44AA99','#882255','#DDCC77', '#117733', '#88CCEE','#999933','#CC6677','#AA4499']
output_loc = 'E:/OceanHealth/output/'
clim=5
data = pd.read_csv(output_loc+ f"results_cat_3_4__"+str(clim)+".csv")
regions = list(data['Region Name'].unique())
regions = ['CCAL','ANTA','PNEC','AUSW','NPTG','SPSG','MONS']
let = ['a','b','c','d','e','f','g','h','i','j','k']
fontsize = 16
# fco2 part
fig, axs = plt.subplots(int(np.ceil(len(regions)/2)), 2, figsize=(18, 1.8 * len(regions)), sharex=False)
axs = axs.ravel()
cou = 0
width = 0.15

for region in regions:
    print(region)
    region_data = data[data['Region Name'] == region]
    
    co2 = region_data[region_data['Parameter'] == 'fgco2']; co2.sort_values(by='months_after', inplace=True)
    wind = region_data[region_data['Parameter'] == 'ERA5_ws2']; wind.sort_values(by='months_after', inplace=True)
    temp = region_data[region_data['Parameter'] == 'temperature']; temp.sort_values(by='months_after', inplace=True)
    sal = region_data[region_data['Parameter'] == 'salinity']; sal.sort_values(by='months_after', inplace=True)
    fco2sw = region_data[region_data['Parameter'] == 'sfco2']; fco2sw.sort_values(by='months_after', inplace=True)
    fco2atm = region_data[region_data['Parameter'] == 'seaflux_fco2_atm']; fco2atm.sort_values(by='months_after', inplace=True)

    wind['Mean Climatology'] = pd.to_numeric(wind['Mean Climatology'])
    temp['Mean Climatology'] = pd.to_numeric(temp['Mean Climatology'])
    sal['Mean Climatology'] = pd.to_numeric(sal['Mean Climatology'])
    fco2sw['Mean Climatology'] = pd.to_numeric(fco2sw['Mean Climatology'])
    fco2atm['Mean Climatology'] = pd.to_numeric(fco2atm['Mean Climatology'])


    base_co2_flux = flux(fco2sw['Mean Climatology'],
                         fco2atm['Mean Climatology'],
                         temp['Mean Climatology'], 
                         sal['Mean Climatology'],
                         wind['Mean Climatology']
                        )
    #print(base_co2_flux)
    fco2sw_co2_flux = flux(pd.to_numeric(fco2sw['Mean Climatology'])+fco2sw['Mean anomaly'],
                         fco2atm['Mean Climatology'],
                         temp['Mean Climatology'], 
                         sal['Mean Climatology'],
                         wind['Mean Climatology']
                        ) - base_co2_flux
    #print(fco2sw_co2_flux)
    fco2atm_co2_flux = flux(fco2sw['Mean Climatology'],
                         pd.to_numeric(fco2atm['Mean Climatology'])+fco2atm['Mean anomaly'],
                         temp['Mean Climatology'], 
                         sal['Mean Climatology'],
                         wind['Mean Climatology']
                        ) - base_co2_flux
    
    temp_co2_flux = flux(fco2sw['Mean Climatology'],
                         fco2atm['Mean Climatology'],
                         pd.to_numeric(temp['Mean Climatology'])+temp['Mean anomaly'], 
                         sal['Mean Climatology'],
                         wind['Mean Climatology']
                        ) - base_co2_flux
    
    sal_co2_flux = flux(fco2sw['Mean Climatology'],
                         fco2atm['Mean Climatology'],
                         temp['Mean Climatology'], 
                         pd.to_numeric(sal['Mean Climatology'])+sal['Mean anomaly'],
                         wind['Mean Climatology']
                        ) - base_co2_flux

    wind_co2_flux = flux(fco2sw['Mean Climatology'],
                         fco2atm['Mean Climatology'],
                         temp['Mean Climatology'], 
                         sal['Mean Climatology'],
                         (pd.to_numeric(wind['Mean Climatology'])+wind['Mean anomaly'])
                        ) - base_co2_flux

    data2 = np.stack((fco2sw_co2_flux,fco2atm_co2_flux,temp_co2_flux,sal_co2_flux,wind_co2_flux))
    cumulated_data = get_cumulated_array(data2, min=0)
    cumulated_data_neg = get_cumulated_array(data2, max=0)
    row_mask = (data2<0)
    cumulated_data[row_mask] = cumulated_data_neg[row_mask]
    data_stack = cumulated_data
    
    for i in np.arange(0, data2.shape[0]):
        axs[cou].bar(np.arange(-12,13), data2[i], bottom=data_stack[i], color=cols[i],width=0.5)
    for i in range(len(wind_co2_flux)):
        if i == 0:
            #plt.bar(i-12-(width*3), sal_run[i]+temp_run[i]+dic_run[i]+ta_run[i],width=width,color=cols[4],label='Residual')
            # axs[cou].bar(i-12-(0.25), np.array(co2['Median anomaly'])[i],width=width,color='k',label='Total')
            axs[cou].bar(i-12-(width*1)-20, fco2sw_co2_flux[i],width=width,color=cols[0],label='fCO$_{2 (sw)}$')
            axs[cou].bar(i-12-(width*0)-20, fco2atm_co2_flux[i],width=width,color=cols[1],label='fCO$_{2 (atm)}$')
            axs[cou].bar(i-12+(width*1)-20, temp_co2_flux[i],width=width,color=cols[2],label='Temperature')
            axs[cou].bar(i-12+(width*2)-20, sal_co2_flux[i],width=width,color=cols[3],label='Salinity')
            axs[cou].bar(i-12+(width*3)-20, wind_co2_flux[i],width=width,color=cols[4],label='Wind')
            # axs[cou].bar(i-12+(width*4)-20, wind_co2_flux[i],width=width,color=cols[5],label='Wind+fCO2sw')
            #axs[cou].bar(i-12+(width*4)-20, wind_fco2a_co2_flux[i],width=width,color=cols[6],label='fCO2atm+wind')
        # else:
        #     axs[cou].bar(i-12-(0.25), np.array(co2['Median anomaly'])[i],width=width,color='k')
    axs[cou].plot([lims[0]-width*3, lims[1]+width*3],[0,0],'k-',linewidth=0.5)
    axs[cou].set_xlim([lims[0]-width*3, lims[1]+width*3])
    # axs[cou].legend()
    axs[cou].grid()
    axs[cou].tick_params(axis='both', which='major', width=1.5, length=6)
    axs[cou].patch.set_visible(False)
    # axs[cou].set_title(region)
   
    axs[cou].text(0.03,0.97,f'({let[cou]}) ' + region,transform=axs[cou].transAxes,va='top',fontweight='bold',fontsize = fontsize)
    
    axs[cou].set_ylabel('Air-sea CO$_2$ flux\nanomaly (mol $m^{-2}$ $s^{-1}$)', fontsize=15)

    cou = cou+1
    
if (len(regions) % 2) != 0:
    axs[-1].remove()
    axs[-2].set_xlabel('Months \n(negative = before heatwave; positive= after heatwave)', fontsize=fontsize)  # Adjust xlabel fontsize
    axs[-3].set_xlabel('Months \n(negative = before heatwave; positive= after heatwave)', fontsize=fontsize)
else:
    axs[-1].set_xlabel('Months \n(negative = before heatwave; positive= after heatwave)', fontsize=fontsize)  # Adjust xlabel fontsize
    axs[-2].set_xlabel('Months \n(negative = before heatwave; positive= after heatwave)', fontsize=fontsize)
plt.subplots_adjust(left=0.1, bottom=0.15, right=0.95, top=0.95, wspace=.3, hspace=.25)
axs[-2].legend(bbox_to_anchor=(1.3,0.75),loc='best',fontsize=20)
   
filename = output_loc+ f'plots/co2flux_decomposition_cat_3_4_plots.png'
fig.savefig(filename)
```
