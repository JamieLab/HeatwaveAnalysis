```python
import numpy as np
import matplotlib.pyplot as plt
import datetime
import geopandas as gpd
from netCDF4 import Dataset
from matplotlib.gridspec import GridSpec
import matplotlib as mpl
from geodatasets import get_path
import matplotlib.transforms

path_to_data = 'C:/Users/df391/AppData/Local/geodatasets/geodatasets/Cache/ne_50m_land/ne_50m_land.shp'
longhurst_file = 'F:/Data/Longhurst/Longhurst_1_deg_old.nc'

c = Dataset(longhurst_file)
longhurst = np.array(c.variables['longhurst'])
lon = np.array(c.variables['lon'])
lat = np.array(c.variables['lat'])
c.close()

provs = np.array([44,38,39,15,2,29,36,53,37,10,22,23,16])
worldmap = gpd.read_file(path_to_data)

for i in range(len(lon)):
    for j in range(len(lat)):
        if longhurst[i,j] not in provs:
            longhurst[i,j] = np.nan
        else:
            f = np.where(longhurst[i,j] == provs)
            longhurst[i,j] = f[0]

font = {'weight' : 'normal',
            'size'   : 16}
matplotlib.rc('font', **font)
fig = plt.figure(figsize=(14,14))
gs = GridSpec(2,1, figure=fig, wspace=0.25,hspace=0.17,bottom=0.05,top=0.97,left=0.1,right=0.95)
ax = [fig.add_subplot(gs[0, 0]),fig.add_subplot(gs[1,0])]

worldmap.plot(color="lightgrey", ax=ax[0],zorder=2)
ax[0].pcolor(lon,lat,np.transpose(longhurst),cmap='tab20b')
ax[0].text(-150,25,'NPTG',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-105,7,'PNEC',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-110,35,'CCAL',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-140,-25,'SPSG',fontsize=16,fontweight='bold',ha='center')
ax[0].text(0,-62,'ANTA',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-15,-25,'SATL',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-55,35,'NWCS',fontsize=16,fontweight='bold',ha='center')
ax[0].text(-25,55,'ARCT',fontsize=16,fontweight='bold',ha='center')
ax[0].text(5,30,'MEDI',fontsize=16,fontweight='bold',ha='center')
ax[0].text(75,-5,'MONS',fontsize=16,fontweight='bold',ha='center')
ax[0].text(75,-25,'ISSG',fontsize=16,fontweight='bold',ha='center')
ax[0].text(128,-30,'AUSW',fontsize=16,fontweight='bold',ha='center')
ax[0].text(160,-45,'TASM',fontsize=16,fontweight='bold',ha='center')
ax[0].text(0.92,0.98,f'(a)',transform=ax[0].transAxes,va='top',fontweight='bold',fontsize = 20)
ax[0].set_xlabel('Longitude ($^o$ E)')
ax[0].set_ylabel('Latitude ($^o$ N)')

heatwave_schematic = 'C:/Users/df391/OneDrive - University of Exeter/Post_Doc_Ocean_Health/Papers/Heatwaves/Heatwave_Schematic.png'
heat = mpl.image.imread(heatwave_schematic)
ax[1].imshow(heat)
ax[1].get_xaxis().set_visible(False)
ax[1].get_yaxis().set_visible(False)
ax[1].text(0.92,0.98,f'(b)',transform=ax[1].transAxes,va='top',fontweight='bold',fontsize = 20)
fig.savefig('E:/OceanHealth/output/plots/Figure_1.png',dpi=300)
```
